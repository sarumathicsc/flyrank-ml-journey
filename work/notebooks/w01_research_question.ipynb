# 2. The Question: Decision, Action, and Cost

## Research Question

Can historical search performance, position stability, and query-level signals help identify content pages that are likely to experience a significant decline in search impressions during a future time window?

## Unit of Analysis

The primary unit of analysis will be a content page at a specific decision point.

## Decision

The decision is which content pages should be reviewed first because they may be at risk of future search-performance decline.

## Output

The final output will be a ranked list of content pages with a predicted decline-risk score and interpretable reason codes explaining the signals behind each recommendation.

## Action

A content or SEO reviewer could investigate the highest-ranked pages and decide whether they require actions such as content updates, search-performance analysis, or further review.

## Cost of a Wrong Recommendation

A false positive could cause a reviewer to spend limited time investigating a page that does not actually decline. A false negative could cause an important declining page to be overlooked.

For this reason, the system should be treated as decision support rather than a guarantee that a page will decline or recover.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/sarumathicsc/flyrank-ml-journey/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

df.head()
```

```python
print("Total content pages:", len(df))

declining = (df["trend_direction"] == "down").sum()
declining_pct = declining / len(df) * 100

print("Pages with declining performance:", declining)
print(f"Percentage of declining pages: {declining_pct:.2f}%")

print("\nAverage impressions_90d by trend direction:")
print(df.groupby("trend_direction")["impressions_90d"].mean().round(2))

print("\nNumber of unique clients:", df["client_id"].nunique())
```

### Observations from the Starter Dataset

The starter dataset contains **30,000 content pages across 32 clients**. I observed that **16,262 pages (54.21%)** are classified with a downward trend, suggesting that search-performance decline is a common enough pattern to investigate.

The average 90-day impressions also differ across trend groups. Pages with a **stable** trend have an average of **9,213.62 impressions**, while pages with a **downward** trend average **4,919.10 impressions**. This observed difference suggests that historical search-performance patterns may contain useful directional signals for identifying pages that could require attention.

These numbers support my provisional **Growth / Recovery / Momentum Prediction** lane. However, these observations do not prove that any individual feature causes a decline. The goal of the project is to provide **decision-support** by identifying potentially at-risk content pages for further review.
