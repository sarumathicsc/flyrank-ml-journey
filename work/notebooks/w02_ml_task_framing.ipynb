### My lane: Predicting content performance decline

I am framing this as a classification problem. The goal is to predict whether a content page is likely to show a decline in search performance based on the signals available about that page.

The prediction can be used to identify pages that may need closer review or a possible content refresh. The model is intended to support prioritization rather than automatically decide which pages should be changed.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/sarumathicsc/flyrank-ml-journey/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

df["target_decline"] = (df["trend_direction"] == "down").astype(int)

df[["content_id", "client_id", "trend_direction", "target_decline"]].head(10)
```

### Target: Content performance decline

For this assignment, I will use `trend_direction == "down"` as a proxy target.

I will create a binary target where 1 means the content page is currently marked as declining and 0 means it is not marked as declining. This gives me a simple target that I can use to demonstrate the classification setup.

This is only a proxy for identifying pages that may need review. It does not mean that a page definitely needs a refresh or that a refresh will improve its performance.

### Success metric

My primary metric will be Precision, because the goal is to prioritize content pages for review. A high precision means that a larger share of the pages flagged as declining are actually marked as declining.

I will also look at Recall and F1-score to make sure the model is not missing too many declining pages and that the overall classification performance is balanced.

For the eventual review workflow, Precision@K can also be useful because the team may only have capacity to review a limited number of pages at a time.

### Why ML beats a fixed rule

A fixed rule could flag every page marked as declining, but it would treat the pages in a similar way without considering the combination of different signals.

An ML model can learn patterns across multiple features, such as impressions, search position, click-through rate, content age, and update history. This can help rank or prioritize pages that are more likely to match the target.

I would still compare the ML model with a simple rule-based baseline. The model should only be useful if it provides better or more useful prioritization than that baseline.
